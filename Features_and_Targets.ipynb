```python
from sklearn.linear_model import LinearRegression

X = [[2], [4], [6], [8]]
y = [60, 70, 80, 90]

model = LinearRegression()

# train model
model.fit(X, y)

# predict score for 10 hours
prediction = model.predict([[10]])

print(prediction)
```
