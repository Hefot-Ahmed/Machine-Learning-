```python
def evaluate_predictions(predictions, correct_answers):

    if len(predictions) != len(correct_answers):
        raise ValueError(
            "Predictions and correct answers must have the same length."
        )

    if len(predictions) == 0:
        raise ValueError(
            "Cannot evaluate an empty list of predictions."
        )

    rewards = []

    for prediction, correct in zip(
        predictions,
        correct_answers
    ):

        reward = reward_function(
            prediction,
            correct
        )

        rewards.append(reward)

    accuracy = sum(rewards) / len(rewards) * 100

    return rewards, accuracy
```

```python
rewards, accuracy = evaluate_predictions(
    model_predictions,
    correct_answers
)

print("Rewards:", rewards)

print(f"Accuracy: {accuracy:.2f}%")
```

```python
import torch
import pandas as pd
import re

from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)

model.eval()

print("Model loaded successfully!")
```

```python
def generate_answer(question):

    messages = [
        {
            "role": "system",
            "content": (
                "You are a mathematics assistant. "
                "Solve the problem carefully. "
                "End your response with 'Final answer: <number>'."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=256,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_tokens = outputs[0][
        inputs["input_ids"].shape[1]:
    ]

    response = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response
```

```python
question = "What is 15 multiplied by 12?"

response = generate_answer(question)

print("Question:")
print(question)

print("\nModel Response:")
print(response)
```

```python
test_dataset = load_dataset(
    "openai/gsm8k",
    "main",
    split="test"
)

test_df = pd.DataFrame(test_dataset)

evaluation_df = test_df.head(10).copy()

print("Number of questions:", len(evaluation_df))

evaluation_df[["question", "answer"]].head()
```

```python
evaluation_df["correct_answer"] = (
    evaluation_df["answer"].apply(extract_answer)
)

evaluation_df["correct_answer_numeric"] = (
    evaluation_df["correct_answer"].apply(convert_to_number)
)

evaluation_df[
    ["question", "correct_answer_numeric"]
].head()
```

```python
model_responses = []

for i, question in enumerate(evaluation_df["question"]):

    print(f"Generating answer {i + 1}/10...")

    response = generate_answer(question)

    model_responses.append(response)

evaluation_df["model_response"] = model_responses

print("All responses generated!")
```

```python
for i in range(3):

    print(f"\nQUESTION {i + 1}")
    print(evaluation_df.iloc[i]["question"])

    print("\nMODEL RESPONSE:")
    print(evaluation_df.iloc[i]["model_response"])

    print("-" * 60)
```

```python
def extract_model_answer(response):

    pattern = (
        r"Final answer:\s*"
        r"([-+]?(?:\d[\d,]*\.?\d*|\.\d+))"
    )

    matches = re.findall(
        pattern,
        response,
        flags=re.IGNORECASE
    )

    if not matches:
        return None

    answer = matches[-1].replace(",", "")

    return float(answer)
```

```python
example_1 = "The answer is 180. Final answer: 180"

example_2 = "Final answer: 1,250"

example_3 = "I don't know."

print(extract_model_answer(example_1))
print(extract_model_answer(example_2))
print(extract_model_answer(example_3))
```

```python
evaluation_df["model_prediction"] = (
    evaluation_df["model_response"].apply(
        extract_model_answer
    )
)

evaluation_df[
    [
        "model_prediction",
        "correct_answer_numeric"
    ]
]
```

```python
def reward_function(prediction, correct_answer):

    if prediction == correct_answer:
        return 1.0

    else:
        return 0.0
```

```python
evaluation_df["reward"] = evaluation_df.apply(
    lambda row: reward_function(
        row["model_prediction"],
        row["correct_answer_numeric"]
    ),
    axis=1
)
```

```python
total_questions = len(evaluation_df)

correct_predictions = int(
    evaluation_df["reward"].sum()
)

accuracy = (
    correct_predictions / total_questions
) * 100

print("=" * 45)
print("REAL AI MODEL EVALUATION")
print("=" * 45)

print("Model:", model_name)

print("Dataset: GSM8K Test")

print("Total Questions:", total_questions)

print("Correct Predictions:", correct_predictions)

print(f"Accuracy: {accuracy:.2f}%")

print("=" * 45)
```

```python
final_results = evaluation_df[
    [
        "question",
        "model_response",
        "model_prediction",
        "correct_answer_numeric",
        "reward"
    ]
].copy()

final_results.columns = [
    "Question",
    "Model Response",
    "Model Prediction",
    "Correct Answer",
    "Reward"
]

final_results
```
